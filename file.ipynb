```python
file_name = "./knowledge_base/sample.txt"

try:
    with open(file_name, "r") as file:
            for line in file:
                print(line.strip())
#except FileNotFoundError:
    #print("File not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
count = 0
#search_name = input("Enter employee name: ")
try:
    with open(file_name, "r") as file:
        for line in file:
            try:
                name, salary, role = line.strip().split(",")
                salary = int(salary)
                
            
                if salary > 70000:
                    print(f"Name: {name}, Salary: {salary}, Role: {role}")
                    count += 1

            except ValueError:
                print(f"Invalid salary value in line: {line.strip()}")
                continue

except FileNotFoundError:
    print("File not found.")
#print(f"Total number of people with search name {search_name} is {count}")
```
