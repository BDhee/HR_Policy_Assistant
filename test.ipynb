```python
from pathlib import Path
def validate_and_get_files(directory_path):
    """ This checks if the directory is valid and returns the list of files in it.

    Args:
        directory_path (_type_): _description_
    """

    if not Path(directory_path).exists():
        raise ValueError(f"The directory {directory_path} does not exist.")

    ##checking if glob got any flies inside the directory
    files = glob.glob(f"{directory_path}/*")


    #checking if files availabvle in the directory
    if not files:
        raise FileNotFoundError(f"No files found in the directory {directory_path}.")
        
    return len(files)
```

```python
knowledge_path = "./knowledge_base"
try:

    file_count = validate_and_get_files(knowledge_path)
    print(f"Number of files in the directory - {file_count}")

    loader = DirectoryLoader(
        path= knowledge_path,
        glob = "**/*.txt",
        loader_cls = TextLoader
    )

    documents= loader.load()

    print(f"Number of documents loaded: {len(documents)} successfully from the directory: {knowledge_path} into langchain.")

except FileNotFoundError as e:
    print(f"File not found error occurred: {e}")
 
```

```python
text_loader = TextLoader(
    file_path="knowledge_base/refunds.txt",
    encoding="utf-8"
)

docs = text_loader.load()

print(f"Number of documents loaded: {len(docs)} successfully from refunds.txt.")
docs
for doc in docs:
    content_preview = doc.page_content[:100]  # Get the first 100 characters
    print(f"Document content preview: {content_preview}...")  # Print the preview
    metadata_preview = doc.metadata  # Get the metadata
    print(f"Document metadata: {metadata_preview}")  # Print the metadata


```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 100,
    length_function = len,
    is_separator_regex = False)

chunks = text_splitter.split_documents(docs)

for i, chunk in enumerate(chunks):
    print(f"Chunk {i}: {chunk.page_content[:100]}...")  # Print the first 100 characters of each chunk  
```
